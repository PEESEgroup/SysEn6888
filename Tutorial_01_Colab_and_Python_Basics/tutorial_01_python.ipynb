## 3. Numbers, strings, and booleans

`int` represents whole numbers, `float` approximates real-valued quantities, `str` stores text, and `bool` is `True` or `False`. Use meaningful names and keep units explicit. `=` assigns a value; `==` compares values.

Division `/` produces a floating-point result; `//` is floor division; `%` gives the remainder; `**` means exponentiation. Floating-point arithmetic is approximate, so compare computed decimal values with a tolerance.

```python
sample_count = 7
label = "Building A"
message = "What's the daily energy use?"  # Double quotes can contain an apostrophe.
is_above_target = energy_kwh > 8
print(type(sample_count).__name__, type(energy_kwh).__name__, type(label).__name__)
print(message, is_above_target)
print("Division:", 7 / 2, "Floor division:", 7 // 2, "Remainder:", 7 % 2, "Square:", 7 ** 2)
print("Exact decimal comparison:", 0.1 + 0.2 == 0.3)
print(label.lower(), label.split())
```

## 4. Choose a collection

| Type | Useful for | Mutable? |
|---|---|---|
| `list` | An ordered sequence of readings | Yes |
| `tuple` | A fixed group such as coordinates | No |
| `dict` | Named values such as units and site | Yes |
| `set` | Unique values and membership checks | Yes |

Python indexing starts at zero; `values[-1]` is the last item. Slicing `values[start:stop]` excludes `stop`. A list assignment can create an alias; use `.copy()` when you want an independent shallow copy.

```python
readings_kwh = [12.0, 9.5, 14.0, 11.0, 8.5]
print("First:", readings_kwh[0], "Last:", readings_kwh[-1], "First three:", readings_kwh[:3])
working = readings_kwh.copy()
working.append(10.0)
removed = working.pop()  # Removes and returns only the last item.
metadata = {"site": "Building A", "unit": "kWh", "synthetic": True}
metadata["samples"] = len(readings_kwh)
coordinates = (42.0, -76.0)  # Invented example, not a measurement location.
unique_sites = {"A", "A", "B"}
print(metadata, coordinates, sorted(unique_sites))
```

## 5. Learn from errors without breaking Run all

An out-of-range index raises `IndexError`; changing a tuple item raises `TypeError`. These are intentional examples, caught with `try`/`except` so the lesson can continue. Catch a specific error you understand instead of hiding every failure.

```python
try:
    print(readings_kwh[99])
except IndexError as error:
    print("Expected IndexError:", error)

try:
    coordinates[0] = 0
except TypeError as error:
    print("Expected TypeError:", error)
```

## 6. Conditions and loops

Use `if` / `elif` / `else` to choose a branch. Indentation defines a block. A `for` loop visits each item; `enumerate` also gives the position. A `while` loop needs a condition that eventually becomes false.

```python
target_kwh = 11.0
for day, reading in enumerate(readings_kwh, start=1):
    if reading > target_kwh:
        status = "above target"
    elif reading == target_kwh:
        status = "at target"
    else:
        status = "below target"
    print(f"Day {day}: {reading:.1f} kWh — {status}")

countdown = 3
while countdown > 0:
    print("Countdown:", countdown)
    countdown -= 1
print("range excludes its stop:", list(range(0, 6, 2)))
```

## 7. Functions and comprehensions

A function packages a calculation with named inputs and a return value. A docstring explains its contract. The caller should know which units and inputs are valid.

List comprehensions combine a simple loop and optional filter. For longer logic, a normal loop is often clearer. `lambda`, `map`, and `filter` are alternatives, not requirements.

```python
def mean_reading(values):
    """Return the arithmetic mean; reject an empty collection."""
    if not values:
        raise ValueError("At least one reading is required.")
    return sum(values) / len(values)

above_target = [value for value in readings_kwh if value > target_kwh]
readings_wh = [1000 * value for value in readings_kwh]
print("Mean:", mean_reading(readings_kwh), "kWh")
print("Above target:", above_target)
print("Same filter:", list(filter(lambda value: value > target_kwh, readings_kwh)))
try:
    mean_reading([])
except ValueError as error:
    print("Expected input validation:", error)
```

## 8. Mini-project: summarize and save measurements

Combine collections, conditions, and functions to create a reusable summary. JSON stores the numbers and their meaning. All inputs here are synthetic; the calculation is real, but it is not evidence about an actual building.

```python
def summarize_energy(values, target):
    return {"count": len(values), "mean_kwh": mean_reading(values),
            "total_kwh": sum(values), "minimum_kwh": min(values), "maximum_kwh": max(values),
            "above_target_count": sum(value > target for value in values),
            "target_kwh": target, "data_status": "synthetic teaching example"}

summary = summarize_energy(readings_kwh, target_kwh)
result_path = OUTPUT_DIR / "energy_summary.json"
result_path.write_text(json.dumps(summary, indent=2))
reloaded = json.loads(result_path.read_text())
print(json.dumps(summary, indent=2))
print("Saved and reloaded:", result_path)
print("TUTORIAL_01_COMPLETE")
```
